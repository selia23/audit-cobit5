```python


import io
import base64
import html
from datetime import date

import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import pandas as pd

# ---------------------------------------------------------------------------
# DATA FORM AUDIT (Tabel 4 - 30 pertanyaan)
# ---------------------------------------------------------------------------

NAMA_PERUSAHAAN = "PT Hartono Istana Teknologi (Polytron), Kudus"
OBJEK_AUDIT = "Sistem informasi penjualan dan persediaan barang (ruang lingkup yang diusulkan)"
KERANGKA_KERJA = "COBIT 5 dengan lima domain proses: EDM, APO, BAI, DSS, dan MEA"
TUJUAN_AUDIT = ("Menilai kecukupan pengendalian atas pengelolaan informasi penjualan dan "
                "persediaan barang berdasarkan lima domain proses COBIT 5, serta "
                "mengidentifikasi area yang perlu diperbaiki.")
RUANG_LINGKUP = ("Pencatatan transaksi penjualan kepada dealer dan pelanggan, penerimaan barang "
                 "jadi dari pabrik ke gudang, pengeluaran dan pengiriman barang, pencatatan "
                 "mutasi dan saldo persediaan, pencocokan data dengan stok fisik, serta "
                 "pelaporan penjualan dan persediaan kepada manajemen.")

NAMA_DOMAIN = {
    'EDM': 'Evaluate, Direct and Monitor (Evaluasi, Pengarahan, dan Pengawasan)',
    'APO': 'Align, Plan and Organise (Perencanaan)',
    'BAI': 'Build, Acquire and Implement (Pembangunan)',
    'DSS': 'Deliver, Service and Support (Pelaksanaan)',
    'MEA': 'Monitor, Evaluate and Assess (Pengawasan)',
}

# (domain, kode proses, pertanyaan audit)
pertanyaan_audit = [
    # ---- EDM ----
    ('EDM', 'EDM01', 'Apakah manajemen melakukan evaluasi berkala terhadap proses pencatatan penjualan dan pengelolaan persediaan barang?'),
    ('EDM', 'EDM01', 'Apakah peran, tanggung jawab, dan kewenangan pengambilan keputusan terkait data penjualan dan persediaan barang di kantor pusat, pabrik, gudang, dan kantor perwakilan telah ditetapkan secara tertulis?'),
    ('EDM', 'EDM02', 'Apakah manajemen mengevaluasi manfaat pengelolaan informasi penjualan dan persediaan terhadap tujuan usaha, misalnya ketepatan pemenuhan pesanan dealer dan efisiensi proses?'),
    ('EDM', 'EDM03', 'Apakah manajemen telah menetapkan batas risiko yang dapat diterima terkait kesalahan, kehilangan, atau perubahan tanpa otorisasi pada data penjualan dan persediaan barang?'),
    ('EDM', 'EDM04', 'Apakah manajemen menyediakan sumber daya yang memadai (pegawai, anggaran, dan sarana pendukung) untuk pengelolaan data penjualan dan persediaan barang?'),
    ('EDM', 'EDM05', 'Apakah laporan penjualan dan persediaan barang disampaikan secara berkala kepada pimpinan dan pihak berkepentingan sesuai ketentuan perusahaan?'),
    # ---- APO ----
    ('APO', 'APO01', 'Apakah perusahaan memiliki prosedur tertulis mengenai pengelolaan data penjualan dan persediaan barang?'),
    ('APO', 'APO01', 'Apakah kebijakan dan prosedur pengelolaan data penjualan dan persediaan telah disosialisasikan kepada pegawai yang berkepentingan?'),
    ('APO', 'APO02', 'Apakah kebutuhan pengelolaan informasi penjualan dan persediaan barang dimuat dalam rencana kerja dan anggaran perusahaan?'),
    ('APO', 'APO07', 'Apakah pegawai yang mengelola data penjualan dan persediaan barang memperoleh pelatihan atau pembekalan yang sesuai dengan tugasnya?'),
    ('APO', 'APO12', 'Apakah perusahaan telah mengidentifikasi dan mendokumentasikan risiko pada proses penjualan dan persediaan barang, seperti selisih stok, kesalahan pencatatan pesanan, atau barang rusak dan tertahan?'),
    ('APO', 'APO13', 'Apakah terdapat kebijakan tertulis mengenai keamanan informasi yang mencakup data penjualan dan persediaan barang?'),
    # ---- BAI ----
    ('BAI', 'BAI02', 'Apakah kebutuhan pengguna (bagian penjualan, gudang, dan keuangan) terhadap pencatatan dan pelaporan penjualan dan persediaan diidentifikasi dan didokumentasikan?'),
    ('BAI', 'BAI04', 'Apakah perusahaan memeriksa kecukupan kapasitas dan ketersediaan sarana pencatatan dibandingkan dengan jumlah transaksi penjualan dan pergerakan persediaan?'),
    ('BAI', 'BAI06', 'Apakah setiap perubahan prosedur atau sarana pencatatan penjualan dan persediaan diajukan dan disetujui secara tertulis oleh pihak berwenang sebelum dilaksanakan?'),
    ('BAI', 'BAI07', 'Apakah perubahan sistem pencatatan penjualan dan persediaan diuji sebelum diterapkan dan hasil pengujiannya didokumentasikan?'),
    ('BAI', 'BAI08', 'Apakah tersedia panduan atau dokumentasi kerja yang dapat diakses pegawai dalam menjalankan pencatatan penjualan dan persediaan barang?'),
    ('BAI', 'BAI09', 'Apakah perusahaan memiliki daftar aset pendukung pengelolaan data penjualan dan persediaan beserta penanggung jawabnya, dan memperbaruinya secara berkala?'),
    # ---- DSS ----
    ('DSS', 'DSS01', 'Apakah kegiatan operasional harian pencatatan penjualan, penerimaan barang jadi, dan pengeluaran barang dilaksanakan sesuai prosedur yang berlaku?'),
    ('DSS', 'DSS02', 'Apakah terdapat prosedur penanganan apabila sistem pencatatan transaksi mengalami gangguan?'),
    ('DSS', 'DSS03', 'Apakah penyebab gangguan atau kesalahan pencatatan yang terjadi berulang dianalisis dan ditindaklanjuti agar tidak terulang?'),
    ('DSS', 'DSS04', 'Apakah terdapat pencadangan data penjualan dan persediaan barang, baik digital maupun dokumen fisik, beserta prosedur pemulihannya?'),
    ('DSS', 'DSS05', 'Apakah akses terhadap data dan catatan penjualan dan persediaan barang dibatasi sesuai kewenangan masing-masing pegawai?'),
    ('DSS', 'DSS06', 'Apakah terdapat pemisahan tugas antara pegawai yang melakukan penjualan, menyimpan dan mengeluarkan barang di gudang, serta mencatat transaksi?'),
    # ---- MEA ----
    ('MEA', 'MEA01', 'Apakah perusahaan menetapkan indikator kinerja proses penjualan dan persediaan barang, misalnya ketepatan pencatatan dan tingkat selisih stok, serta memantaunya secara berkala?'),
    ('MEA', 'MEA01', 'Apakah capaian penjualan dan pengelolaan persediaan dibandingkan dengan target perusahaan, dan penyimpangannya dianalisis serta dilaporkan kepada manajemen?'),
    ('MEA', 'MEA02', 'Apakah perusahaan melakukan pemeriksaan berkala (stok opname) terhadap kesesuaian data persediaan dengan stok fisik di gudang?'),
    ('MEA', 'MEA02', 'Apakah terdapat fungsi pengawasan intern atau fungsi sejenis yang menilai pengendalian pada proses penjualan dan persediaan barang?'),
    ('MEA', 'MEA02', 'Apakah kelemahan pengendalian yang ditemukan dari pemeriksaan internal maupun eksternal ditindaklanjuti dan statusnya dipantau?'),
    ('MEA', 'MEA03', 'Apakah perusahaan memantau kepatuhan proses penjualan dan persediaan barang terhadap ketentuan eksternal, misalnya ketentuan perpajakan atau perjanjian dengan dealer dan pihak ketiga?'),
]

DAFTAR_PERTANYAAN = [
    {"no": i, "domain": d, "kode": k, "teks": t}
    for i, (d, k, t) in enumerate(pertanyaan_audit, start=1)
]
TOTAL = len(DAFTAR_PERTANYAAN)          # 30
URUTAN_DOMAIN = ['EDM', 'APO', 'BAI', 'DSS', 'MEA']
PILIHAN = ['Ya', 'Tidak', 'N/A']

# ---------------------------------------------------------------------------
# STATE
# ---------------------------------------------------------------------------

posisi = 0
jawaban = [None] * TOTAL     # tiap elemen: {'jawaban','bukti','catatan'}
identitas = {}
NAMA_FILE = 'Hasil_Audit_COBIT5_Polytron.xlsx'

# ---------------------------------------------------------------------------
# KOMPONEN UI
# ---------------------------------------------------------------------------

STYLE = """
<style>
    .ausi h3, .ausi h4 { margin: 0 0 5px 0; }
    .ausi p { margin: 0 0 10px 0; }
    .ausi table { border-collapse: collapse; }
    .ausi th, .ausi td { border: 1px solid #999; padding: 6px 8px; }
    .ausi th { background: #f2f2f2; }
</style>
"""

header_label = widgets.HTML(value=STYLE + f"""
<div class='ausi' style='margin-bottom: 15px;'>
    <h3>FORM AUDIT SISTEM INFORMASI PENJUALAN DAN PERSEDIAAN BARANG</h3>
    <h3>{html.escape(NAMA_PERUSAHAAN)} - COBIT 5</h3>
    <b>Petunjuk pengisian jawaban:</b><br>
    <b>Ya</b> - Pengendalian diterapkan dan terdapat bukti yang mendukung.<br>
    <b>Tidak</b> - Pengendalian tidak diterapkan atau tidak dapat dibuktikan.<br>
    <b>N/A</b> - Pertanyaan tidak relevan dengan kondisi objek audit.<br>
    <i>Jawaban "Ya" hanya diberikan apabila bukti pendukung telah diperiksa.</i>
</div>
""")

# --- Halaman identitas (Tabel 3) ---
LAYOUT_FIELD = widgets.Layout(width='600px')
STYLE_FIELD = {'description_width': '150px'}

in_tanggal = widgets.DatePicker(description='Tanggal audit:', value=date.today(),
                                layout=LAYOUT_FIELD, style=STYLE_FIELD)
in_auditor = widgets.Text(description='Nama auditor:', layout=LAYOUT_FIELD, style=STYLE_FIELD)
in_responden = widgets.Text(description='Responden:', layout=LAYOUT_FIELD, style=STYLE_FIELD)
in_bagian = widgets.Text(description='Bagian/unit kerja:', layout=LAYOUT_FIELD, style=STYLE_FIELD)
btn_mulai = widgets.Button(description='Mulai Audit', button_style='primary', icon='play',
                           layout=widgets.Layout(width='160px', margin='10px 0 0 0'))
pesan_identitas = widgets.HTML()

identitas_info = widgets.HTML(value=f"""
<div class='ausi'>
    <h4>Identitas Audit</h4>
    <table>
      <tr><th>Nama perusahaan</th><td>{html.escape(NAMA_PERUSAHAAN)}</td></tr>
      <tr><th>Objek audit</th><td>{html.escape(OBJEK_AUDIT)}</td></tr>
      <tr><th>Kerangka kerja</th><td>{html.escape(KERANGKA_KERJA)}</td></tr>
    </table>
    <p style='margin-top:10px'><i>Isi identitas berikut sebelum pemeriksaan dimulai.</i></p>
</div>
""")

halaman_identitas = widgets.VBox([
    identitas_info, in_tanggal, in_auditor, in_responden, in_bagian, btn_mulai, pesan_identitas
])

# --- Halaman pertanyaan ---
progres = widgets.IntProgress(value=0, min=0, max=TOTAL, bar_style='info',
                              layout=widgets.Layout(width='400px'))
progres_label = widgets.HTML()
judul_label = widgets.HTML()
pertanyaan_label = widgets.HTML()
input_jawaban = widgets.ToggleButtons(options=PILIHAN, value=None, description='Jawaban:',
                                      button_style='info', style={'description_width': '80px'})
input_bukti = widgets.Textarea(description='Bukti pendukung:',
                               placeholder='Dokumen, catatan transaksi, SOP, tangkapan layar, atau hasil pemeriksaan',
                               layout=widgets.Layout(width='700px', height='70px'),
                               style={'description_width': '120px'})
input_catatan = widgets.Textarea(description='Catatan auditor:',
                                 placeholder='Penjelasan mengenai kondisi yang ditemukan',
                                 layout=widgets.Layout(width='700px', height='70px'),
                                 style={'description_width': '120px'})
pesan_error = widgets.HTML()
btn_kembali = widgets.Button(description='Kembali', icon='arrow-left',
                             layout=widgets.Layout(width='130px'))
btn_lanjut = widgets.Button(description='Simpan & Lanjut', button_style='primary',
                            icon='arrow-right', layout=widgets.Layout(width='170px'))

halaman_form = widgets.VBox([
    widgets.HBox([progres, progres_label]),
    judul_label, pertanyaan_label,
    input_jawaban, input_bukti, input_catatan, pesan_error,
    widgets.HBox([btn_kembali, btn_lanjut]),
])

# --- Halaman hasil ---
tombol_ekspor = widgets.Button(description=' Ekspor ke Excel', button_style='success',
                               icon='file-excel-o',
                               layout=widgets.Layout(margin='20px 10px 10px 0', width='200px'))
tombol_ubah = widgets.Button(description='Ubah Jawaban', icon='edit',
                             layout=widgets.Layout(margin='20px 0 10px 0', width='170px'))
pesan_ekspor = widgets.HTML()
hasil_html = widgets.HTML()
halaman_hasil = widgets.VBox([hasil_html, widgets.HBox([tombol_ekspor, tombol_ubah]), pesan_ekspor])

root = widgets.VBox([header_label, halaman_identitas])


# ---------------------------------------------------------------------------
# LOGIKA REKAPITULASI
# ---------------------------------------------------------------------------

def hitung_rekap():
    """Mengembalikan (DataFrame Tabel 6, total_dict)."""
    baris = []
    tot = {'Jumlah Pertanyaan': 0, 'Ya': 0, 'Tidak': 0, 'N/A': 0}
    for dom in URUTAN_DOMAIN:
        idx = [i for i, q in enumerate(DAFTAR_PERTANYAAN) if q['domain'] == dom]
        ya = sum(1 for i in idx if jawaban[i] and jawaban[i]['jawaban'] == 'Ya')
        tidak = sum(1 for i in idx if jawaban[i] and jawaban[i]['jawaban'] == 'Tidak')
        na = sum(1 for i in idx if jawaban[i] and jawaban[i]['jawaban'] == 'N/A')
        baris.append({'Domain': dom, 'Jumlah Pertanyaan': len(idx), 'Ya': ya,
                      'Tidak': tidak, 'N/A': na,
                      'Indikator Kepatuhan (%)': indikator(ya, len(idx), na)})
        tot['Jumlah Pertanyaan'] += len(idx)
        tot['Ya'] += ya
        tot['Tidak'] += tidak
        tot['N/A'] += na
    baris.append({'Domain': 'Total', **tot,
                  'Indikator Kepatuhan (%)': indikator(tot['Ya'], tot['Jumlah Pertanyaan'], tot['N/A'])})
    return pd.DataFrame(baris), tot


def indikator(ya, jumlah, na):
    """Indikator kepatuhan sederhana = Ya / (Jumlah Pertanyaan - N/A) x 100%."""
    penyebut = jumlah - na
    if penyebut <= 0:
        return None
    return round(ya / penyebut * 100, 2)


def fmt_persen(v):
    return '-' if v is None or pd.isna(v) else f"{v:.2f}%"


def hitung_temuan():
    """Daftar pertanyaan berjawaban 'Tidak' sebagai bahan awal Tabel 7."""
    hasil = []
    for i, q in enumerate(DAFTAR_PERTANYAAN):
        j = jawaban[i]
        if j and j['jawaban'] == 'Tidak':
            hasil.append({
                'No.': len(hasil) + 1,
                'Ref. No.': q['no'],
                'Temuan Audit': j['catatan'] or 'Pengendalian tidak diterapkan atau tidak dapat dibuktikan.',
                'Risiko yang Mungkin Timbul': '',
                'Rekomendasi Perbaikan': '',
                'Penanggung Jawab': '',
                'Tindak Lanjut': '',
            })
    return hasil


# ---------------------------------------------------------------------------
# LOGIKA TAMPILAN
# ---------------------------------------------------------------------------

def mulai_audit(b=None):
    global identitas
    nama = in_auditor.value.strip()
    resp = in_responden.value.strip()
    bagian = in_bagian.value.strip()
    if not (nama and resp and bagian and in_tanggal.value):
        pesan_identitas.value = ("<span style='color:red'><b>Tanggal audit, nama auditor, responden, "
                                 "dan bagian/unit kerja wajib diisi sebelum audit dimulai.</b></span>")
        return
    pesan_identitas.value = ""
    identitas = {
        'Nama perusahaan': NAMA_PERUSAHAAN,
        'Objek audit': OBJEK_AUDIT,
        'Kerangka kerja audit': KERANGKA_KERJA,
        'Tanggal audit': in_tanggal.value.strftime('%d-%m-%Y'),
        'Nama auditor': nama,
        'Responden': resp,
        'Bagian atau unit kerja': bagian,
        'Tujuan audit': TUJUAN_AUDIT,
        'Ruang lingkup audit': RUANG_LINGKUP,
    }
    root.children = [header_label, halaman_form]
    tampilkan_pertanyaan()


def tampilkan_pertanyaan():
    q = DAFTAR_PERTANYAAN[posisi]
    progres.value = posisi
    progres_label.value = f"&nbsp;Pertanyaan {posisi + 1} dari {TOTAL}"
    judul_label.value = (f"<div class='ausi'><b>DOMAIN {q['domain']} - {NAMA_DOMAIN[q['domain']]}"
                         f"<br>Kode proses: {q['kode']}</b></div>")
    pertanyaan_label.value = f"<div class='ausi'><p><b>{q['no']}.</b> {html.escape(q['teks'])}</p></div>"

    j = jawaban[posisi]
    input_jawaban.value = j['jawaban'] if j else None
    input_bukti.value = j['bukti'] if j else ''
    input_catatan.value = j['catatan'] if j else ''
    pesan_error.value = ""
    btn_kembali.disabled = (posisi == 0)
    btn_lanjut.description = 'Simpan & Selesai' if posisi == TOTAL - 1 else 'Simpan & Lanjut'


def simpan_dan_lanjut(b=None):
    global posisi
    pilihan = input_jawaban.value
    bukti = input_bukti.value.strip()
    catatan = input_catatan.value.strip()

    if pilihan is None:
        pesan_error.value = "<span style='color:red'><b>Pilih jawaban Ya, Tidak, atau N/A terlebih dahulu.</b></span>"
        return
    if pilihan == 'Ya' and not bukti:
        pesan_error.value = ("<span style='color:red'><b>Jawaban \"Ya\" hanya boleh diberikan jika bukti "
                             "pendukung sudah diperiksa. Isi kolom Bukti pendukung, atau ubah jawaban.</b></span>")
        return

    jawaban[posisi] = {'jawaban': pilihan, 'bukti': bukti, 'catatan': catatan}
    pesan_error.value = ""

    if posisi < TOTAL - 1:
        posisi += 1
        tampilkan_pertanyaan()
    else:
        tampilkan_hasil()


def kembali(b=None):
    global posisi
    if posisi > 0:
        posisi -= 1
        tampilkan_pertanyaan()


def ubah_jawaban(b=None):
    global posisi
    posisi = TOTAL - 1
    root.children = [header_label, halaman_form]
    tampilkan_pertanyaan()


def grafik_base64(df_rekap):
    d = df_rekap[df_rekap['Domain'] != 'Total']
    fig, ax = plt.subplots(figsize=(8, 4))
    bawah = [0] * len(d)
    for kolom, warna in [('Ya', '#4CAF50'), ('Tidak', '#E57373'), ('N/A', '#BDBDBD')]:
        nilai = d[kolom].tolist()
        ax.bar(d['Domain'], nilai, bottom=bawah, color=warna, edgecolor='black', label=kolom)
        for x, (v, b0) in enumerate(zip(nilai, bawah)):
            if v > 0:
                ax.text(x, b0 + v / 2, str(v), ha='center', va='center', fontweight='bold')
        bawah = [b0 + v for b0, v in zip(bawah, nilai)]
    ax.set_ylim(0, 7)
    ax.set_ylabel('Jumlah Pertanyaan')
    ax.set_title('Rekapitulasi Jawaban per Domain COBIT 5', fontweight='bold')
    ax.grid(axis='y', linestyle='--', alpha=0.7)
    ax.legend(loc='upper right', ncol=3)
    buf = io.BytesIO()
    plt.savefig(buf, format='png', bbox_inches='tight')
    plt.close(fig)
    return base64.b64encode(buf.getvalue()).decode('utf-8')


def tampilkan_hasil():
    df_rekap, tot = hitung_rekap()
    img = grafik_base64(df_rekap)

    h = "<div class='ausi'><h2>Pengisian Form Audit Selesai</h2>"
    h += "<h3>1. Grafik Rekapitulasi per Domain</h3>"
    h += f'<img src="data:image/png;base64,{img}" alt="Grafik rekapitulasi" style="margin-bottom:20px;"/>'

    h += "<h3>2. Rekapitulasi Jawaban Form Audit (Tabel 6)</h3>"
    h += ("<table style='width:100%; text-align:left; margin-bottom:10px;'><tr><th>Domain</th>"
          "<th>Jumlah Pertanyaan</th><th>Ya</th><th>Tidak</th><th>N/A</th>"
          "<th>Indikator Kepatuhan</th></tr>")
    for _, r in df_rekap.iterrows():
        tebal = r['Domain'] == 'Total'
        o, c = ('<b>', '</b>') if tebal else ('', '')
        h += (f"<tr><td>{o}{r['Domain']}{c}</td><td>{o}{r['Jumlah Pertanyaan']}{c}</td>"
              f"<td>{o}{r['Ya']}{c}</td><td>{o}{r['Tidak']}{c}</td><td>{o}{r['N/A']}{c}</td>"
              f"<td>{o}{fmt_persen(r['Indikator Kepatuhan (%)'])}{c}</td></tr>")
    h += "</table>"
    h += ("<p><i>Indikator kepatuhan sederhana = Jumlah Ya ÷ (Jumlah Pertanyaan − Jumlah N/A) × 100%. "
          "Nilai ini hanya indikator berdasarkan jawaban kuesioner dan bukan nilai capability level "
          "resmi COBIT 5.</i></p>")

    temuan = hitung_temuan()
    h += "<h3>3. Pertanyaan Berjawaban \"Tidak\" (bahan awal Tabel 7)</h3>"
    if temuan:
        h += ("<table style='width:100%; text-align:left; margin-bottom:20px;'><tr><th>Ref. No.</th>"
              "<th>Kode</th><th>Pertanyaan Audit</th><th>Catatan Auditor</th></tr>")
        for t in temuan:
            q = DAFTAR_PERTANYAAN[t['Ref. No.'] - 1]
            cat = jawaban[t['Ref. No.'] - 1]['catatan']
            h += (f"<tr><td>{q['no']}</td><td>{q['kode']}</td><td>{html.escape(q['teks'])}</td>"
                  f"<td>{html.escape(cat) or '-'}</td></tr>")
        h += "</table>"
    else:
        h += "<p>Tidak ada pertanyaan yang dijawab \"Tidak\".</p>"
    h += "</div>"

    hasil_html.value = h
    pesan_ekspor.value = ""
    root.children = [header_label, halaman_hasil]


# ---------------------------------------------------------------------------
# EKSPOR EXCEL
# ---------------------------------------------------------------------------

def _format_sheet(ws, lebar):
    from openpyxl.styles import Alignment, Font, PatternFill, Border, Side
    tipis = Side(style='thin', color='999999')
    garis = Border(left=tipis, right=tipis, top=tipis, bottom=tipis)
    for kol, w in lebar.items():
        ws.column_dimensions[kol].width = w
    for row in ws.iter_rows():
        for c in row:
            c.alignment = Alignment(wrap_text=True, vertical='top')
            c.border = garis
    for c in ws[1]:
        c.font = Font(bold=True)
        c.fill = PatternFill('solid', fgColor='F2F2F2')
        c.alignment = Alignment(wrap_text=True, vertical='center', horizontal='center')
    ws.freeze_panes = 'A2'


def ekspor_ke_excel(b=None):
    try:
        # Sheet 1: Identitas (Tabel 3)
        df_identitas = pd.DataFrame(list(identitas.items()), columns=['Komponen', 'Keterangan'])

        # Sheet 2: Form audit (Tabel 4)
        baris = []
        for i, q in enumerate(DAFTAR_PERTANYAAN):
            j = jawaban[i] or {'jawaban': '', 'bukti': '', 'catatan': ''}
            baris.append({
                'No.': q['no'], 'Domain': q['domain'], 'Kode Proses': q['kode'],
                'Pertanyaan Audit': q['teks'],
                'Ya': '✓' if j['jawaban'] == 'Ya' else '',
                'Tidak': '✓' if j['jawaban'] == 'Tidak' else '',
                'N/A': '✓' if j['jawaban'] == 'N/A' else '',
                'Bukti Pendukung': j['bukti'], 'Catatan Auditor': j['catatan'],
            })
        df_form = pd.DataFrame(baris)

        # Sheet 3: Rekapitulasi (Tabel 6)
        df_rekap, _ = hitung_rekap()
        df_rekap = df_rekap.copy()
        df_rekap['Catatan'] = ''
        df_rekap['Indikator Kepatuhan (%)'] = df_rekap['Indikator Kepatuhan (%)'].apply(
            lambda v: '-' if v is None or pd.isna(v) else v)

        # Sheet 4: Temuan (Tabel 7)
        temuan = hitung_temuan()
        df_temuan = pd.DataFrame(temuan) if temuan else pd.DataFrame(
            columns=['No.', 'Ref. No.', 'Temuan Audit', 'Risiko yang Mungkin Timbul',
                     'Rekomendasi Perbaikan', 'Penanggung Jawab', 'Tindak Lanjut'])

        with pd.ExcelWriter(NAMA_FILE, engine='openpyxl') as writer:
            df_identitas.to_excel(writer, sheet_name='Identitas Audit', index=False)
            df_form.to_excel(writer, sheet_name='Form Audit', index=False)
            df_rekap.to_excel(writer, sheet_name='Rekapitulasi', index=False)
            df_temuan.to_excel(writer, sheet_name='Temuan', index=False)

            _format_sheet(writer.sheets['Identitas Audit'], {'A': 26, 'B': 90})
            _format_sheet(writer.sheets['Form Audit'],
                          {'A': 6, 'B': 9, 'C': 12, 'D': 70, 'E': 6, 'F': 8, 'G': 6, 'H': 40, 'I': 40})
            _format_sheet(writer.sheets['Rekapitulasi'],
                          {'A': 12, 'B': 20, 'C': 8, 'D': 8, 'E': 8, 'F': 26, 'G': 30})
            _format_sheet(writer.sheets['Temuan'],
                          {'A': 6, 'B': 9, 'C': 45, 'D': 30, 'E': 30, 'F': 20, 'G': 20})

        pesan_ekspor.value = ("<div style='color: green; padding: 10px; border: 1px solid green; "
                              f"border-radius: 5px;'><b>✅ Berhasil!</b> File disimpan di folder ini "
                              f"dengan nama: <b>{NAMA_FILE}</b></div>")
        try:  # otomatis mengunduh jika dijalankan di Google Colab
            from google.colab import files
            files.download(NAMA_FILE)
        except Exception:
            pass
    except Exception as e:
        pesan_ekspor.value = f"<div style='color: red;'><b>Gagal mengekspor:</b> {html.escape(str(e))}</div>"


# ---------------------------------------------------------------------------
# INISIALISASI
# ---------------------------------------------------------------------------

btn_mulai.on_click(mulai_audit)
btn_lanjut.on_click(simpan_dan_lanjut)
btn_kembali.on_click(kembali)
tombol_ekspor.on_click(ekspor_ke_excel)
tombol_ubah.on_click(ubah_jawaban)

display(root)
```
